# Exercice 03 — Transposée d'une matrice

La transposée de $A$, notée $A^T$, s'obtient en échangeant les **lignes et les colonnes**.

$$ A = \begin{pmatrix}1&2&3\\4&5&6\end{pmatrix}_{2\times3}
\quad\Longrightarrow\quad
A^T = \begin{pmatrix}1&4\\2&5\\3&6\end{pmatrix}_{3\times2} $$

**Règle mathématique :** $(A^T)_{ij}=A_{ji}$. Il n'est pas nécessaire que $A$ soit carrée.

**Objectif :** saisir une matrice $n\times m$, calculer sa transposée **sans NumPy**, puis vérifier avec NumPy. Pour chaque notion, les commentaires du code expliquent les usages concrets dans la vie quotidienne, l'informatique, l'IA et l'architecture.


```python
import math

# ============================================================
# EXERCICE 03 — TRANSPOSÉE D'UNE MATRICE : PYTHON PUR
# ============================================================
#
# 1) MATHÉMATIQUES
# Une matrice A de taille n x m possède n lignes et m colonnes.
# Sa transposée AT a m lignes et n colonnes.
# Formule : AT[j][i] = A[i][j]
# Il ne faut NI additionner NI multiplier les coefficients :
# on les replace simplement à une nouvelle position.
# Exemple : A = [[1, 2, 3], [4, 5, 6]]
#           AT = [[1, 4], [2, 5], [3, 6]]
# Propriété à vérifier : transposer(transposer(A)) == A.
#
# 2) VIE QUOTIDIENNE
# Une entreprise peut représenter des ventes avec :
#   lignes = magasins ; colonnes = produits.
# La transposée organise la même information autrement :
#   lignes = produits ; colonnes = magasins.
# ATTENTION : transposer ne calcule pas de nouvelles ventes.
#
# 3) INFORMATIQUE / DONNÉES
# Un tableau n x m stocke n observations et m caractéristiques.
# Avec la transposée m x n, on regarde les observations
# regroupées par caractéristique plutôt que l'inverse.
# Cette réorganisation est utile en analyse de données.
#
# 4) INTELLIGENCE ARTIFICIELLE / MACHINE LEARNING
# Pour une matrice de données X (n exemples x m variables),
# X.T intervient dans certains calculs de régression linéaire.
# Par exemple : X.T @ (X @ w - y) intervient dans le gradient
# de l'erreur quadratique (à un facteur de normalisation près).
# La transposition rend ici possibles les dimensions du produit.
#
# 5) ARCHITECTURE / GÉNIE CIVIL
# On peut stocker les coordonnées 3D de n points d'un bâtiment :
#   n lignes (points) x 3 colonnes (x, y, z).
# Après transposition : 3 lignes (axes) x n colonnes (points).
# Cela facilite certaines opérations de géométrie matricielle.
# Important : la transposée ne change PAS la géométrie du
# bâtiment et ne réalise PAS une rotation à elle seule.
#
# 6) ALGORITHME
# Deux boucles : i parcourt les lignes, j les colonnes.
# La case A[i][j] est copiée en AT[j][i].
# ============================================================


def lire_entier_positif(message):
    """Demander un nombre de lignes ou de colonnes strictement positif."""
    while True:
        try:
            valeur = int(input(message))
            if valeur > 0:
                return valeur
        except ValueError:
            pass
        print("Veuillez saisir un entier strictement positif.")


def lire_matrice(nom, n, m):
    """Lire n lignes de m nombres finis, séparés par des espaces."""
    matrice = []
    print(f"\nSaisie de {nom} ({n} x {m})")
    for i in range(n):
        while True:
            entree = input(f"Ligne {i + 1} ({m} valeurs séparées par des espaces) : ")
            try:
                ligne = [float(v) for v in entree.split()]
            except ValueError:
                print("Valeurs invalides. Utilisez des nombres, séparés par des espaces.")
                continue
            if len(ligne) != m or not all(math.isfinite(v) for v in ligne):
                print(f"Saisissez exactement {m} nombres finis.")
                continue
            matrice.append(ligne)
            break
    return matrice


def transposer(A):
    """Transposer une matrice rectangulaire non vide, sans NumPy."""
    if not A or not A[0]:
        raise ValueError("La matrice ne doit pas être vide.")
    n, m = len(A), len(A[0])
    if any(len(ligne) != m for ligne in A):
        raise ValueError("Chaque ligne doit avoir le même nombre de colonnes.")

    AT = [[0.0 for _ in range(n)] for _ in range(m)]

    for i in range(n):
        # i = indice de ligne dans la matrice A.
        for j in range(m):
            # j = indice de colonne dans A.
            # Le coefficient à (i, j) passe à (j, i).
            # Exemple : A[0][2] = 3 devient AT[2][0] = 3.
            AT[j][i] = A[i][j]
    return AT


def afficher_matrice(nom, A):
    print(f"\n{nom} ({len(A)} x {len(A[0])}) =")
    for ligne in A:
        print("[" + "  ".join(f"{v:g}" for v in ligne) + "]")
```

```python
# ============================================================
# SAISIE INTERACTIVE : UNE SEULE MATRICE A
# A peut être carrée OU rectangulaire (n x m).
# ============================================================
print("=== EXERCICE 03 : TRANSPOSÉE D'UNE MATRICE ===")
n = lire_entier_positif("Nombre de lignes n : ")
m = lire_entier_positif("Nombre de colonnes m : ")

A = lire_matrice("A", n, m)
AT = transposer(A)

afficher_matrice("A", A)
afficher_matrice("Transposée Aᵀ", AT)

# Double transposée : elle doit redonner la matrice d'origine.
assert transposer(AT) == A
print("\nVérification (Aᵀ)ᵀ = A : OK")
```

```python
# ============================================================
# VÉRIFICATION NUMPY : on compare avec A_numpy.T
# IMPORTANT : NumPy vérifie notre code, il ne le remplace pas.
# ============================================================
import numpy as np

A_numpy = np.array(A)
AT_numpy = A_numpy.T

print("Transposée avec NumPy :")
print(AT_numpy)
print("Identique à notre algorithme :", np.array_equal(AT_numpy, np.array(AT)))
assert np.array_equal(AT_numpy, np.array(AT))
```

```python
# ============================================================
# TESTS AUTOMATIQUES : pas de saisie nécessaire.
# ============================================================
assert transposer([[1, 2, 3], [4, 5, 6]]) == [[1, 4], [2, 5], [3, 6]]
assert transposer([[1, 2], [3, 4]]) == [[1, 3], [2, 4]]
assert transposer([[1, 2, 3]]) == [[1], [2], [3]]
assert transposer([[1], [2], [3]]) == [[1, 2, 3]]
assert transposer(transposer([[2, 4, 6], [8, 10, 12]])) == [[2, 4, 6], [8, 10, 12]]

for invalide in ([], [[]], [[1, 2], [3]]):
    try:
        transposer(invalide)
    except ValueError:
        pass
    else:
        raise AssertionError("Une matrice invalide aurait dû être refusée.")

print("Tous les tests Python ont réussi !")
```
