```python
# ============================================================
# EXERCICE 01 — ADDITION DE MATRICES
# ============================================================
#
# 1. DÉFINITION MATHÉMATIQUE
#
# Une matrice est un tableau de nombres organisé en lignes
# et en colonnes.
#
# Pour additionner deux matrices A et B, elles doivent
# posséder exactement les mêmes dimensions.
#
# Si A et B sont de dimension n × m, alors :
#
#              C[i][j] = A[i][j] + B[i][j]
#
# On additionne les éléments qui occupent la même position.
#
# Exemple :
#
#     A = [[1, 2],       B = [[5, 6],
#          [3, 4]]            [7, 8]]
#
#     A + B = [[6, 8],
#              [10, 12]]
#
#
# 2. APPLICATION DANS LA VIE QUOTIDIENNE
#
# Imaginons une entreprise qui possède deux magasins.
#
# Chaque ligne représente un magasin et chaque colonne
# représente un produit.
#
# Matrice A : ventes du lundi
#
#     A = [[10, 20],
#          [15, 25]]
#
# Matrice B : ventes du mardi
#
#     B = [[5, 8],
#          [6, 10]]
#
# En additionnant A et B :
#
#     C = [[15, 28],
#          [21, 35]]
#
# On obtient les quantités vendues sur les deux journées.
#
# Cette opération permet d'agréger facilement des données.
#
#
# 3. APPLICATION EN INFORMATIQUE
#
# Les images numériques sont représentées par des tableaux
# de valeurs numériques.
#
# Dans une image en niveaux de gris, chaque élément peut
# représenter l'intensité d'un pixel.
#
# Une matrice A peut représenter une image.
# Une matrice B peut représenter une correction d'intensité.
#
# Additionner A + B permet de modifier les pixels.
#
# Attention : pour une image classique à 8 bits,
# les valeurs finales doivent rester entre 0 et 255.
# Il faut donc parfois limiter les résultats.
#
#
# 4. APPLICATION EN INTELLIGENCE ARTIFICIELLE
#
# Les réseaux de neurones effectuent des opérations
# mathématiques sur des matrices et des vecteurs.
#
# Une opération courante est :
#
#                 Z = W × X + B
#
# W : poids du réseau
# X : données d'entrée
# B : biais
# Z : résultat avant la fonction d'activation
#
# L'addition permet d'ajouter les biais au résultat
# de la multiplication matricielle.
#
# Dans certains frameworks, les biais sont ajoutés
# par broadcasting (extension automatique des dimensions).
#
#
# 5. APPLICATION EN ARCHITECTURE ET GÉNIE CIVIL
#
# Lorsqu'on étudie un bâtiment, on peut organiser
# certaines charges par étage et par zone.
#
# Exemple pédagogique :
#
# A : charges permanentes en kN
#
#     A = [[10, 20],
#          [15, 25]]
#
# B : charges d'exploitation en kN
#
#     B = [[5, 8],
#          [6, 10]]
#
# Chaque position correspond à une même zone du bâtiment.
#
# Le total simple est :
#
#     C = [[15, 28],
#          [21, 35]]
#
# L'addition permet de combiner les deux catégories.
#
# IMPORTANT :
# Ce calcul est une illustration mathématique.
# Le dimensionnement réel exige des combinaisons
# de charges réglementaires et une analyse structurelle.
#
#
# 6. OBJECTIF DE NOTRE PROGRAMME
#
# Notre programme doit :
#
# - Accepter une matrice A de dimension n × m.
# - Accepter éventuellement une deuxième matrice B.
# - Si une seule matrice est saisie, calculer A + A.
# - Si deux matrices sont saisies, calculer A + B.
# - Parcourir les lignes et colonnes.
# - Calculer chaque élément de la matrice résultat.
# - Afficher le résultat.
#
# ============================================================
```

```python
import math

def lire_entier(message, minimum=1, maximum=None):
    while True:
        try:
            valeur = int(input(message))

            if valeur < minimum:
                print(f"Entrez un entier >= {minimum}.")
                continue

            if maximum is not None and valeur > maximum:
                print(f"Entrez un entier <= {maximum}.")
                continue

            return valeur

        except ValueError:
            print("Veuillez entrer un entier valide.")


def lire_matrice(nom, lignes, colonnes):
    matrice = []

    print(f"\nSaisie de la matrice {nom}")

    for i in range(lignes):
        while True:
            try:
                saisie = input(
                    f"Ligne {i + 1} ({colonnes} valeurs séparées par des espaces) : "
                )

                valeurs = [
                    float(nombre)
                    for nombre in saisie.split()
                ]

                if len(valeurs) != colonnes:
                    print(f"Il faut exactement {colonnes} valeurs.")
                    continue

                if not all(math.isfinite(v) for v in valeurs):
                    print("Entrez uniquement des nombres finis.")
                    continue

                matrice.append(valeurs)
                break

            except ValueError:
                print("Valeurs invalides. Réessayez.")

    return matrice


def addition_matrices(A, B):
    if len(A) != len(B):
        raise ValueError("Nombre de lignes différent.")

    if not A or not B:
        raise ValueError("Les matrices ne doivent pas être vides.")

    colonnes = len(A[0])

    if colonnes == 0 or any(
        len(ligne) != colonnes for ligne in A + B
    ):
        raise ValueError("Dimensions invalides.")

    C = []

    for i in range(len(A)):
        ligne = [] # On parcourt chaque ligne de la matrice.

        for j in range(colonnes):
            
        # i représente l'indice de la ligne.
        # j représente l'indice de la colonne.
        #
        # Exemple dans le cas des ventes :
        # A[i][j] = quantité vendue lundi.
        # B[i][j] = quantité vendue mardi.
        #
        # La somme représente la quantité totale
        # vendue pour ce magasin et ce produit.
        
            somme = A[i][j] + B[i][j]
            ligne.append(somme)

        C.append(ligne)

    return C


def afficher_matrice(nom, matrice):
    print(f"\n{nom} =")

    for ligne in matrice:
        print(ligne)
```

```python
print("=== ADDITION DE MATRICES ===")

mode = lire_entier(
    "Nombre de matrices à saisir (1 ou 2) : ",
    minimum=1,
    maximum=2
)

n = lire_entier("Nombre de lignes : ")
m = lire_entier("Nombre de colonnes : ")

A = lire_matrice("A", n, m)

if mode == 1:
    B = [ligne.copy() for ligne in A]
else:
    B = lire_matrice("B", n, m)

C = addition_matrices(A, B)

afficher_matrice("A", A)
afficher_matrice("B", B)
afficher_matrice("Résultat C", C)
```

```python
import numpy as np

A_numpy = np.array(A)
B_numpy = np.array(B)

C_numpy = A_numpy + B_numpy

print("Résultat NumPy :")
print(C_numpy)

print(
    "Résultats identiques :",
    np.allclose(C_numpy, C)
)
```
