## Solving by substitution — Pure Python

We first solve a 2×2 linear system without NumPy.

The general system is:

\[
a_1x + b_1y = c_1
\]

\[
a_2x + b_2y = c_2
\]

The user enters the coefficients of the two equations.

The program then applies the substitution method automatically.

```python
def solve_by_substitution():
    print("Solve a system of the form:")
    print("a1*x + b1*y = c1")
    print("a2*x + b2*y = c2")
    print()

    # First equation
    a1 = float(input("a1 = "))
    b1 = float(input("b1 = "))
    c1 = float(input("c1 = "))

    print()

    # Second equation
    a2 = float(input("a2 = "))
    b2 = float(input("b2 = "))
    c2 = float(input("c2 = "))

    print("\nSystem:")
    print(f"{a1}x + ({b1})y = {c1}")
    print(f"{a2}x + ({b2})y = {c2}")

    # We isolate y from the first equation:
    #
    # a1*x + b1*y = c1
    #
    # b1*y = c1 - a1*x
    #
    # y = (c1 - a1*x) / b1

    if b1 == 0:
        print("\nCannot isolate y from the first equation.")
        return

    print("\n--- Substitution method ---")

    print(
        f"From equation 1:\n"
        f"y = ({c1} - {a1}x) / {b1}"
    )

    # Substitute y into equation 2:
    #
    # a2*x + b2*((c1 - a1*x)/b1) = c2

    denominator = a2 - (b2 * a1 / b1)
    numerator = c2 - (b2 * c1 / b1)

    if denominator == 0:
        if numerator == 0:
            print("\nThe system has infinitely many solutions.")
        else:
            print("\nThe system has no solution.")
        return

    x = numerator / denominator

    # Now substitute x back into equation 1
    y = (c1 - a1 * x) / b1

    print("\nAfter substitution:")
    print(f"x = {x}")

    print("\nSubstitute x back into equation 1:")
    print(f"y = ({c1} - {a1} × {x}) / {b1}")
    print(f"y = {y}")

    print("\n--- Solution ---")
    print(f"x = {x}")
    print(f"y = {y}")

    # Verification
    equation_1 = a1 * x + b1 * y
    equation_2 = a2 * x + b2 * y

    print("\n--- Verification ---")
    print(f"Equation 1: {equation_1} = {c1}")
    print(f"Equation 2: {equation_2} = {c2}")
```

```python
solve_by_substitution()
```

```python

# Matrix A: coefficients of x and y
A = np.array([
    [2, 1],
    [1, -1]
], dtype=float)

# Vector b: values on the right side
b = np.array([5, 1], dtype=float)

# Solve Ax = b
solution = np.linalg.solve(A, b)

x, y = solution

print("x =", x)
print("y =", y)
```

```python

def solve_by_substitution_numpy(A, b):
    a1, b1 = A[0]
    a2, b2 = A[1]

    c1, c2 = b

    if np.isclose(b1, 0):
        raise ValueError(
            "Cannot isolate y from the first equation."
        )

    denominator = a2 - (b2 * a1 / b1)
    numerator = c2 - (b2 * c1 / b1)

    if np.isclose(denominator, 0):
        if np.isclose(numerator, 0):
            return "Infinitely many solutions"
        else:
            return "No solution"

    x = numerator / denominator
    y = (c1 - a1 * x) / b1

    return np.array([x, y])
```

```python
# Usage
A = np.array([
    [2, 1],
    [1, -1]
], dtype=float)

b = np.array([5, 1], dtype=float)

solution = solve_by_substitution_numpy(A, b)

print("Solution:", solution)
```
